```python
import torch
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
```

```python
inp = torch.linspace(-3,3,1000)
# print(inp)
dx = inp[1] - inp[0] # 0.667
# print(torch.diff(inp))
rout = F.relu(inp)
gout = F.gelu(inp)
# print(rout)
dif_relu = torch.diff(rout)
d_relu = dif_relu/dx
# print(dif_relu)
# print(dif_relu/dx)
dif_gelu = torch.diff(gout)
d_gelu = dif_gelu/dx
# print(d_gelu.shape)

_,axs =  plt.subplots(1,2, figsize=(15,4))
axs[0].plot(inp, rout, label="ReLU")
axs[0].plot(inp, gout, label="GeLU")
axs[0].set(ylabel = "output value")
axs[0].legend()

axs[1].plot(inp[:-1], d_relu, label="D ReLU")
axs[1].plot(inp[:-1], d_gelu, label="D GeLU")
axs[1].set(ylabel = "Derivative value")
axs[1].legend()
```
