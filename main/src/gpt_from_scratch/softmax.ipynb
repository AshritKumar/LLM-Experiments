```python
import torch
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-2, 2, 0.2)
print(x.shape)

_, axs = plt.subplots(1,2, figsize=(15, 4))

# T = 0.5
# sf = np.exp(x/T)/np.sum(np.exp(x/T))
# axs[0].plot(x, sf, '--', c='pink', label="T=0.5")
# axs[1].plot(x, np.log(sf), '--', c='pink', label="T=0.5")

T = 1
sf = np.exp(x/T)/np.sum(np.exp(x/T))
# print(sf.sum())

axs[0].plot(x, sf, 'o-', c='black', label="T=1")
axs[1].plot(x, np.log(sf), 'o-', c='black', label="T=1")

T = 1.2
sf = np.exp(x/T)/np.sum(np.exp(x/T))
axs[0].plot(x, sf, 'o-', c='red', label="T=1.2")
axs[1].plot(x, np.log(sf), 'o-', c='red', label="T=1.2")

T = 1.5
sf = np.exp(x/T)/np.sum(np.exp(x/T))
axs[0].plot(x, sf, 'o-', c='blue', label="T=1.5")
axs[1].plot(x, np.log(sf), 'o-', c='blue', label="T=1.5")

T = 2
sf = np.exp(x/T)/np.sum(np.exp(x/T))
axs[0].plot(x, sf, 'o-', c='green', label="T=2")
axs[1].plot(x, np.log(sf), 'o-', c='green', label="T=2")

axs[0].legend()
axs[1].legend()
plt.show()
```
