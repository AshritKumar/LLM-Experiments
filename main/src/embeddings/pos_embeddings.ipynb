```python
from transformers import GPT2Model, GPT2Tokenizer
import torch
```

```python
gpt2 = GPT2Model.from_pretrained("gpt2")
gpt2_tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
```

```python
gtp2E = gpt2.wte.weight.detach()
gpt2PE = gpt2.wpe.weight.detach()
```

```python
# gpt2PE[:,0].shape
# cosine sim b/w every embedding

import matplotlib.pyplot as plt

peMagnitudes = torch.norm(gpt2PE, dim=1, keepdim=True) # 1024,1 => for each embedding across all 1024 tokens [POSITION OR TIME]
peNoms = gpt2PE / peMagnitudes
# peNoms.shape
peCosSims = peNoms @ peNoms.T
plt.imshow(peCosSims)
plt.colorbar()
plt.show()
```
