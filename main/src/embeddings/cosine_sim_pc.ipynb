```python
import numpy as np
```

```python
a = np.array([32, 4, 55])
b = np.array([11, 9, 65])

# cosine similarity = a.b / (|a| * |b|)

print(sum(a*b)) # a.b
Sc = sum(a*b) / (np.sqrt(sum(a*a)) * np.sqrt(sum(b*b)))
print(f"Manual cosine similarity: {Sc}")

# Pearson correlation

aMean = np.mean(a)
bMean = np.mean(b)

num = sum((a-aMean)*(b-bMean))
denom = np.sqrt(sum((a-aMean)**2)) * np.sqrt(sum((b-bMean)**2))
Pc = num/denom

print(f"Manual Pearson correlation: {Pc}")

```

```python
import torch

at = torch.tensor([1, 2, 3], dtype=torch.float32)
bt = torch.tensor([4, 5, 6],  dtype=torch.float32)
av = at.view(1,-1)
bv = bt.view(1,-1)
Sct = torch.cosine_similarity(av, bv, dim=1)
print("Cosine similarity: ", Sct)

print("Pearson correlation: ", np.corrcoef(a, b)[0,1])
```
